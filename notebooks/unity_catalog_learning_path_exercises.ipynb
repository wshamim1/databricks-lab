## Setup Variables

Adjust these values before running SQL cells.

```python
catalog_name = "retail_uc_demo"
schema_name = "curated"
table_name = "orders_summary"
analyst_group = "retail_analysts"

print(catalog_name, schema_name, table_name)
```

## Exercise 1: Create Catalog and Schema

Use Unity Catalog hierarchy: `catalog.schema.table`.

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_name}")
spark.sql("SHOW CATALOGS").show(truncate=False)
spark.sql(f"SHOW SCHEMAS IN {catalog_name}").show(truncate=False)
```

## Exercise 2: Create and Populate a Managed Table

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog_name}.{schema_name}.{table_name} (
  order_date DATE,
  orders_count BIGINT,
  gross_revenue DECIMAL(18,2)
)
""")

spark.sql(f"""
INSERT INTO {catalog_name}.{schema_name}.{table_name}
VALUES
  (current_date(), 42, 12050.00),
  (date_sub(current_date(), 1), 39, 10120.50)
""")

spark.sql(f"SELECT * FROM {catalog_name}.{schema_name}.{table_name} ORDER BY order_date DESC").show()
```

## Exercise 3: Apply and Inspect Grants

If your identity has permission-grant privileges, run these statements.

```python
grant_sql = f"GRANT SELECT ON TABLE {catalog_name}.{schema_name}.{table_name} TO `{analyst_group}`"
print("Run grant manually if needed:", grant_sql)

spark.sql(f"SHOW GRANTS ON TABLE {catalog_name}.{schema_name}.{table_name}").show(truncate=False)
```

## Exercise 4: Troubleshooting Checks

Run these checks when access errors occur.

```python
spark.sql("SELECT current_user() AS active_principal").show(truncate=False)
spark.sql(f"DESCRIBE TABLE EXTENDED {catalog_name}.{schema_name}.{table_name}").show(truncate=False)
spark.sql(f"SHOW TABLES IN {catalog_name}.{schema_name}").show(truncate=False)
```
